```python
def comment_rating_counter(srcs  ,pattern = r"star-fill\.svg$") : 
    try : 
        star_fill_urls = [src for src in srcs if re.search(pattern, src)]
        return len(star_fill_urls)
    except : 
        pass
```

```python
def get_stars_pics_srcs(comment) :
    try : 
        text_and_comment_box = comment.find_element(By.CSS_SELECTOR , 'div.flex.gap-1.flex-col')
        comment_text = text_and_comment_box.find_element(By.CSS_SELECTOR , 'p.text-body-1.text-neutral-900.mb-1.break-words').text
        rating_box = text_and_comment_box.find_elements(By.CSS_SELECTOR , 'img')
        src = []
        for i in rating_box : 
            src.append(i.get_attribute('src'))
        return [comment_text , src]
    except : 
        pass
```

```python
def continue_button(comment) : 
    text_and_comment_box = comment.find_element(By.CSS_SELECTOR , 'div.flex.gap-1.flex-col')
    try : 
        continue_button = text_and_comment_box.find_element(By.CSS_SELECTOR , "span.inline-flex.items-center.cursor-pointer.styles_Anchor--secondary__3KsgY.text-button-2.mt-2 span")
        driver.execute_script("arguments[0].click();" , continue_button)
    except NoSuchElementException as e : 
        print('Here is the error continue button')
```

```python
def color_and_votes(comment) : 
    color_and_vote = comment.find_element(By.CSS_SELECTOR , 'div.w-full.flex.flex-row.mt-2.justify-between.items-center')
    color = color_and_vote.find_element(By.CSS_SELECTOR , 'p.text-caption.text-neutral-600.inline.ml-1').text
    votes_box = color_and_vote.find_element(By.CSS_SELECTOR , 'div.mr-auto.lg\\:mr-0.flex.items-center')
    like_and_dislike_box = votes_box.find_elements(By.CSS_SELECTOR , 'button.relative.flex.items-center.user-select-none.styles_btn__Q4MvL.text-button-2.rounded-medium.px-2.text-neutral-650')
    likes = like_and_dislike_box[0].find_element(By.CSS_SELECTOR , 'p.lg\\:text-caption.text-caption-compact').text
    dislikes = like_and_dislike_box[1].find_element(By.CSS_SELECTOR , 'p.lg\\:text-caption.text-caption-compact').text
    return [color , likes , dislikes]
```

```python
def comments_per_page() : 
    last_height = driver.execute_script("return document.body.scrollHeight")
    while True:
        driver.execute_script('window.scrollTo(0, document.body.scrollHeight)')
        time.sleep(1)  # کمتر از قبل برای سرعت؛ اگر لازم دیدی افزایش بده
        new_height = driver.execute_script("return document.body.scrollHeight")
        if new_height == last_height:
            break
        last_height = new_height
    print('scrolling completed')


    comment_section = loading.find_element(By.ID , 'commentSection')
    driver.execute_script("arguments[0].scrollIntoView();", comment_section)
    print('comment section')



    try : 
        more_comments = driver.find_elements(By.CLASS_NAME, "border-complete-t-200.py-3")
        for item in more_comments : 
            if "دیدگاه" in item.text and "دیگر" in item.text :
                btn = item.find_element(By.CSS_SELECTOR , 'button')
                btn.click()
    except NoSuchElementException as e : 
        print('Here is the error more comments')


    comments = driver.find_elements(By.CSS_SELECTOR , "div.grow.min-w-0 article[class='lg:mt-0 py-3 br-list-vertical-no-padding-200']")
    print('commesnt list')
    comments_lists = []
    for comment in comments : 
        comment_dict = {}
        continue_button(comment)
        print('continue button')
        time.sleep(0.5)
        srcs = get_stars_pics_srcs(comment)
        print('srcs')
        time.sleep(0.5)
        text = srcs[0]
        rating = comment_rating_counter(srcs[1])
        print('comment rating counter')
        time.sleep(0.5)
        color = color_and_votes(comment)[0]
        likes = color_and_votes(comment)[1]
        dislikes = color_and_votes(comment)[2]
        print('color and votes')
        
        # print(comment_text , '\n' * 5)

        comment_dict['text'] = text
        comment_dict['rating'] = rating
        comment_dict['color'] = color
        comment_dict['likes'] = likes
        comment_dict['dislikes'] = dislikes
        comments_lists.append(comment_dict)

    return comments_lists
        
    
```
